```python
 # import libraries 
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns 
import plotly.express as px
import plotly.io as pio

pio.renderers.default = "svg"
```

```python

# importing the dataset
raw_df = pd.read_csv("Zomato Chennai Listing 2020.csv")
raw_df.head()
```

## cleaning the dataset

### remove redundunt rows data

```python
# dropping columns which are not required for further analysis
raw_df.drop(['Zomato URL','Address'],axis=1, inplace=True)
raw_df.head()
```

###  removing the null records

```python
# checking for the null records
raw_df.isnull().sum()
```

```python
# Replacing the other null records with NA
raw_df.fillna(0,inplace=True)
```

```python
# confirming all the null records are correct 
raw_df.isnull().sum()
```

###  Converting the Datatypes of numerical columns to numeric datatype

```python
raw_df['Dining Rating Count'].value_counts()
```

```python
raw_df['Dining Rating Count'] = raw_df['Dining Rating Count'].replace(to_replace=['Does not offer Dining','Not enough Dining Reviews'],value=0)
```

```python
raw_df['Delivery Rating Count'].value_counts()
```

```python
raw_df['Delivery Rating Count'] = raw_df['Delivery Rating Count'].replace(to_replace=['Does not offer Delivery','Not enough Delivery Reviews'],value=0)
```

```python
# changing data type of the numerical columns
raw_df['Dining Rating Count']=raw_df['Dining Rating Count'].astype('int64')
raw_df['Delivery Rating Count']=raw_df['Delivery Rating Count'].astype('int64')
```

```python
raw_df.info()
```

###  Removing the restaurant records whose Rating or Votes is 0

```python
# finding those restaurant whose has 0 Rating or Votes 
useless_data = ((raw_df['Dining Rating']==0.0) &  (raw_df['Dining Rating Count']==0)) & ((raw_df['Delivery Rating']==0.0) &  (raw_df['Delivery Rating Count']==0))
raw_df[useless_data]
```

```python
# performing negation of the useless dataset and then storing the correct data back in the raw_df DataFrame
# this permanently remove the wrong data from the original dataframe
raw_df=raw_df[~useless_data]
```

```python
for i in raw_df.columns:
    print(f"{i} \n\n {raw_df[i].value_counts()} \n\n\n")
```

###  Working on 'Name' column

```python
raw_df['Name of Restaurant'].value_counts()
```

```python
# Removing the irrelevant text from the region column 
raw_df['Name of Restaurant'] = raw_df['Name of Restaurant'].str.replace('[a-zA-Z].+-','',regex=True)
raw_df['Name of Restaurant']=raw_df['Name of Restaurant'].str.strip()
```

```python
raw_df['Name of Restaurant'].value_counts()
```

###  Working on Top Dishes column 

```python
raw_df['Top Dishes'].value_counts()
```

```python
raw_df['Top Dishes'] = raw_df['Top Dishes'].replace('Invalid','Not Mentioned')
raw_df['Top Dishes'].value_counts()
```

###  Working on price column 

```python
# convert price for 2 into price for 1
raw_df['Price for 2'] = raw_df['Price for 2']/2
```

```python
raw_df.rename(columns={'Price for 2':'PRICE'},inplace=True)
```

```python
raw_df
```

###  removing Duplicate records

```python
# finding all duplicate records
raw_df[raw_df.duplicated()]    # No duplicate records
```

###  copying the cleaned data into a new DataFrame

```python
zomato_chennai_df = raw_df.copy()
zomato_chennai_df.head()
```

```python
zomato_chennai_df.info()
```

# Performing Exploratory Data Analysis

## 1. how many restaurants are Top 25 highly rated (dining) for briyani in chennai?

```python
briyani_df = zomato_chennai_df[zomato_chennai_df['Cuisine'].str.contains('Biryani')]
briyani_df = briyani_df.sort_values(by='Dining Rating',ascending=False).head(25)
briyani_df
```

```python
fig=px.histogram(briyani_df,x='Name of Restaurant',color='PRICE',
                title='Top 25 highly rated (dining) restaurants for briyani in chennai').update_xaxes(categoryorder="total descending")
fig.show()
```

## 2. how the price for each cusine differs by location?    -- sort by Dining Rating

```python
cusine_price_df = zomato_chennai_df.groupby(['Cuisine','Location','Dining Rating'])['PRICE'].mean().reset_index()
cusine_price_df = cusine_price_df[cusine_price_df['Dining Rating']>4.5]
cusine_price_df
```

```python
fig=px.scatter(cusine_price_df,x='Location',y='Cuisine',color='PRICE',
              title="Average price for each cusine based on location",
              width=800,   # Width in pixels
              height=600 )
fig.show()
```

## 3. How many top dishs are available for each cusine in restaurant

```python
# top_dish_df = zomato_chennai_df.groupby(['Cuisine','Dining Rating','Top Dishes'])['PRICE'].mean().reset_index()
# top_dish_df = top_dish_df[top_dish_df['Dining Rating']>4.5]
# top_dish_df

top_dish_df = zomato_chennai_df.groupby(['Cuisine','Name of Restaurant'])['Top Dishes'].count().reset_index()
top_dish_df.sort_values(by='Top Dishes',ascending=False)
```

## 4. high dining and delivery rated restaurants 

```python
highrate_rest_df = zomato_chennai_df[['Name of Restaurant','Location','Cuisine','PRICE','Features','Dining Rating','Delivery Rating']]
highdine_df = highrate_rest_df[highrate_rest_df['Dining Rating']>4.7]
highdine_df
```

```python
highdeli_df = highrate_rest_df[highrate_rest_df['Delivery Rating']>4.5]
highdeli_df
```

## how many restaurants are in chennai for each type of cuisine?

```python
chennai_rest_df = zomato_chennai_df.groupby('Cuisine')['Name of Restaurant'].count().reset_index()
chennai_rest_df.sort_values(by='Name of Restaurant',ascending=False)
```

## which areas have a large Number of Desserts restaurant Market ?

```python
desert_df = zomato_chennai_df[zomato_chennai_df['Cuisine'].str.contains('Desserts')]
desert_df = desert_df[desert_df['Dining Rating']>4.0]
desert_df
```

```python
desertrest_df = desert_df.groupby(by='Location').agg({'Name of Restaurant':'count','PRICE':'mean'}).reset_index().rename(columns={'Name of Restaurant':'Count Of Restaurants'})
desertrest_df = desertrest_df.sort_values('Count Of Restaurants',ascending=False).head(25)
desertrest_df.head()
```

```python
fig=px.bar(desertrest_df,x='Location' ,y='Count Of Restaurants' ,color='PRICE' ,
           title='no of desserts restaurant by places',height=700)
fig.show()
```

## Is there a relation between Price and region

```python
region_price_df = zomato_chennai_df.groupby(['Location'])['PRICE'].mean().reset_index()
region_price_df
```

```python
fig=px.scatter(region_price_df,x='Location' ,y='PRICE',height=700).update_traces(marker_size=7)
fig.show()
```

## list of affordable restaurants

```python
zomato_chennai_df.head()
```

```python
max_price=zomato_chennai_df['PRICE'].max()
one_fourth_price=max_price/4
one_fourth_price
```

```python
aff_rest_df=zomato_chennai_df[['Name of Restaurant','PRICE','Cuisine','Location','Top Dishes']]
aff_rest_df = aff_rest_df[aff_rest_df['PRICE']<=625]
aff_rest_df.sort_values(by='PRICE',inplace=True)
aff_rest_df
```

```python
 # Finding the highest rated list of restaurants 
highrate_rest_df=zomato_chennai_df[['Name of Restaurant','PRICE','Cuisine','Location','Top Dishes','Dining Rating']]
highrate_rest_df = highrate_rest_df[highrate_rest_df['Dining Rating']>=4.5]
highrate_rest_df.sort_values(by='PRICE',inplace=True)
highrate_rest_df
```

```python
highrate_aff_df = pd.merge(aff_rest_df,highrate_rest_df,how='inner',on=['Name of Restaurant','Location'])
highrate_aff_df=highrate_aff_df[['Name of Restaurant','PRICE_x','Cuisine_x','Location','Top Dishes_x']]
highrate_aff_df.rename(columns={'Name of Restaurant':'Name of Restaurant','PRICE_x':'PRICE','Cuisine_x':'Cuisine',
                                'Location':'Location','Top Dishes_x':'Top Dishes'},inplace=True)
```
